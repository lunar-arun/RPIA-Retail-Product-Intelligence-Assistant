### Objective: Clean your dataset and make it ready for NLP analysis and vector storage.

1. **Load your dataset**

```python
import pandas as pd

df = pd.read_csv("data/raw/fake_reviews.csv")
print(df.head())
```

2. **Basic Cleaning**

   * Drop missing rows:

```python
df.dropna(subset=['review_text', 'product_name'], inplace=True)
```

   * Convert column names to lowercase:

```python
df.columns = df.columns.str.lower()
```

3. **Save Cleaned Data**

   Save your cleaned version to:

```python
df.to_csv("data/processed/clean_reviews.csv", index=False)
```
